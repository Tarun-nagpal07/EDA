## City Day Univariate

```python
city_day.columns
```

```python
sns.lineplot(x='AQI',y='Date',data=city_day)
```

```python
cols = ['PM2.5', 'PM10', 'NO', 'NO2', 'NOx', 'NH3', 'CO', 'SO2',
       'O3', 'Benzene', 'Toluene', 'Xylene']

for col in cols:
    plt.Figure(figsize=(6,10))
    sns.histplot(city_day[col])
    plt.show()
```

```python
for col in cols:
    plt.Figure(figsize=(6,10))
    sns.boxplot(city_day[col])
    plt.show()
```

```python
city_day['PM2.5'].skew()
```

**PM2.5 is highly right-skewed with many extreme outliers,
indicating occasional severe pollution events.
The median (~45) is much lower than the extreme values
reaching above 900**

```python
pollutants = ['PM2.5','PM10','NO2','SO2','CO','O3']

for col in pollutants:
    print(col, city_day[(city_day[col] < 0) | (city_day[col] > 1000)].shape[0])
```

```python
for col in cols:
    city_day[col] = np.log1p(city_day[col])
```

```python
fig, ax = plt.subplots(4,3, figsize=(15,12))

for i, col in enumerate(cols):
    
    row = i // 3
    col_pos = i % 3
    
    sns.histplot(city_day[col], ax=ax[row, col_pos])
    ax[row, col_pos].set_title(col)

plt.tight_layout()
plt.show()
```

```python
fig , ax = plt.subplots(4,3,figsize=(15,12))

for i, col in enumerate(cols):
    
    row = i // 3
    col_pos = i % 3
    
    sns.scatterplot(x=col,y='AQI',data=city_day, ax=ax[row, col_pos])
    ax[row, col_pos].set_title(col)

plt.tight_layout()
plt.show()
```

PM2.5 and PM10 exhibit the strongest relationships with AQI,
confirming that particulate matter is the primary driver of air pollution.

Nitrogen-based pollutants (NO, NO2, NOx) also show strong positive trends.

SO2, NH3, and O3 have moderate influence on AQI levels.

Volatile organic compounds such as Benzene, Toluene, and Xylene
display weak relationships with AQI.

Overall, AQI appears to increase significantly when multiple pollutants
rise simultaneously.
